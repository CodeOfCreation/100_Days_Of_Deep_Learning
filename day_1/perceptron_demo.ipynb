```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('placement (2).csv')
df.shape
df.head()
```

```python
sns.scatterplot(df['cgpa'],df['resume_score'],hue=df['placed'])
```

```python
X = df.iloc[:,0:2]
y = df.iloc[:,-1]
```

```python
from sklearn.linear_model import Perceptron
model = Perceptron()
```

```python
model.fit(X,y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions

plot_decision_regions(X.values, y.values, clf=model, legend=2)
```
